## Intro to Tensors

### Creating Tensors

```python
# scalar
scalar = torch.tensor(3)
scalar
```

```python
# but I wanted that number '3' not like 'tensor(3)'

scalar.item()  # .item() method to get the python number from a tensor containing a single value
```

```python
# dim of an scalar
scalar.ndim
```

```python
print(f"Hence an Scalar has {scalar.ndim} dimension.")
```

```python
#  vector

vector = torch.tensor([11, 14, 1])
vector
```

```python
# Again i wanted that number '11' and '2' not like 'tensor([11,  2])'

ls = vector.tolist()  # .tolist() method to get the python list from a tensor containing multiple values
print(f"Vector as list: {ls}")
# or 

np_array = vector.numpy()  # .numpy() method to get the numpy array from a tensor containing multiple values
print(f"Vector as numpy array: {np_array}, Array contains no comma.")

# or

vector[0].item()  # .item() method to get the python number from a tensor containing a single value
vector[1].item()  # .item() method to get the python number from a tensor containing a single value
```

```python
for i in range(len(vector)):
    print(vector[i].item())
```

```python
# one thing to see
vector.ndim
```

```python
print(f"Hence a Vector has {vector.ndim} dimension.")
```

**Well those Square Bracket counts specify the number of the dim**

```python
# BUT
print(f"Vector shape: {vector.shape}")
# number of elements in the vector
print(f"Number of elements in the vector: {vector.numel()}")
```

### Creating a range of tensors and tensors-like

```python
one_to_ten = torch.arange(start = 1, end = 11, step = 1.5)
one_to_ten
# returns 1-D range of numbers from  ceil((end-start)/step) 
# Step is added to the start value until it reaches the end
```

```python
np.ceil((11-1)/1.5)
```

```python
one_to_ten.size()
```

```python
# tensor - like

ten_zeros = torch.zeros_like(input = one_to_ten)
ten_zeros
```

```python
ten_ones = torch.ones_like(input = one_to_ten)
ten_ones
```

### Tensor data types

```python
# float 32 tensor
float_32_tensor = torch.tensor([1.0, 2.0, 3.0],
                               dtype=None,
                               device=None,
                               requires_grad=False 
# requires_grad is used to track the gradients of the tensor during backpropagation
                               )
float_32_tensor.dtype
```

```python
float_16_tensor = torch.tensor([1.0, 2.0, 3.0], dtype=torch.float16)
float_16_tensor.dtype
```

Conversion

```python
# or
new_float_16_tensor = float_32_tensor.type(torch.float16)
new_float_16_tensor.dtype
```

OR

```python
new2_float_16_tensor = float_32_tensor.to(torch.float16)
new2_float_16_tensor.dtype
```

```python
float_64_tensor = torch.tensor([111.0, 222.0, 333.0], dtype=torch.float64)
```

```python
n = (float_16_tensor * float_32_tensor * float_64_tensor)
```

```python
n.dtype
```
