```python
# one thing to see
vector.ndim
```

**Well those Square Bracket counts specify the number of the dim**

#### Matrix Multiplication

***Rules***:
1. `(3, 2) @ (3, 2)` ❌
2. `(2, 3) @ (3, 2)` ✅ -> `(2, 2)`
3. `(3, 2) @ (2, 3)` ✅ -> `(3, 3)`


```python
print("Element-wise multiplication:\n",tensor, "*", tensor, "=", tensor * tensor)
```

```python
print("matrix multiplication:\n",tensor, "@", tensor, "=", tensor @ tensor)
```

or

```python
print("matrix multiplication:\n",tensor, "@", tensor, "=", torch.matmul(tensor, tensor))
```

#### Difference between these

```python
%%time

value = 0

for i in range(len(tensor)):
    value += tensor[i] * tensor[i]
```

```python
%%time
tensor @ tensor
```

```python
%%time
torch.matmul(tensor, tensor)
```

`matmul()` 29.1 μs < `@` 37 μs < `for loop` 193 μs

***Recommended To Use:*** 
`torch.matmul()`
