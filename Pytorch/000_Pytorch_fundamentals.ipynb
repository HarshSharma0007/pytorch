#### Matrix Multiplication

***Rules***:
1. `(3, 2) @ (3, 2)` ❌
2. `(2, 3) @ (3, 2)` ✅ -> `(2, 2)`
3. `(3, 2) @ (2, 3)` ✅ -> `(3, 3)`


```python
print("Element-wise multiplication:\n",tensor, "*", tensor, "=", tensor * tensor)
```

```python
print("matrix multiplication:\n",tensor, "@", tensor, "=", tensor @ tensor)
```

or

```python
print("matrix multiplication:\n",tensor, "@", tensor, "=", torch.matmul(tensor, tensor))
```

#### Difference between these

```python
%%time

value = 0

for i in range(len(tensor)):
    value += tensor[i] * tensor[i]
```

```python
%%time
tensor @ tensor
```

```python
%%time
torch.matmul(tensor, tensor)
```

`matmul()` 29.1 μs < `@` 37 μs < `for loop` 193 μs

***Recommended To Use:*** 
`torch.matmul()`

Above error states at the end, that:

`Got: Long` which is actually `float64` or the `Double-precision floating-point format`

but we can update it to the `Single-precision floating-point format` which is actually `float32`

```python

# Convert the tensor to float32 before calculating the mean
torch.mean(x.type(torch.float32)), x.type(torch.float32).mean()
```

***torch.reshape***

```python
x = torch.arange(1. , 10.)
x, x.shape
```

```python
# Add an extra dim
# x_reshaped = x.reshape(1, 7) # shape should be multipe of the original input size


x_reshaped = x.reshape(1, 9)
x_reshaped, x_reshaped.shape
```

***torch.view***

```python
# change the view
z = x.view(1, 9)
z, z.shape
```

***View***: Changing z changes x. Cause they share the same memory!!

```python
z[:, 0] = 5
z, x, z == x
```

```python
# x[0] = 1
# x, z
```

***torch.stack***
https://docs.pytorch.org/docs/2.11/generated/torch.stack.html

```python
# Stack tensors on top of eachother

x_stacked = torch.stack(
    [x, x, x, x],
    dim = 0,     
    # dim = 0 or -2 : Horizontal
    # dim = 1 or -1 : Vertical
)
x_stacked
```

***Squeeze*** : Returns tensor with all the dimensions of input of size 1 removed

https://docs.pytorch.org/docs/2.11/generated/torch.squeeze.html



```python
x_reshaped.shape
```

```python
x_reshaped.shape
```

Apply the squeeze function to remove the `1` dimension from x_reshaped

```python
torch.squeeze(x_reshaped, 0) ,torch.squeeze(x_reshaped, 0).shape
```

```python
x_reshaped.shape
```

Since we had, 1x9, and applied the dim = 0 we got the 1 removed,

```python
torch.squeeze(x_reshaped, 1) , torch.squeeze(x_reshaped, 1).shape
```

So better way is also:

```python
torch.squeeze(x_reshaped, (1,0)) , torch.squeeze(x_reshaped, (1,0)).shape
```

```python
print(f"Previous Tensor: {x_reshaped}")
print(f"Previous Shape: {x_reshaped.shape}")

# remove extra dimensions

x_squeezed = x_reshaped.squeeze()
print(f"\nNew tensor: {x_squeezed}")
print(f"New shape: {x_squeezed.shape}")
```

***Unsqueeze***
https://docs.pytorch.org/docs/2.11/generated/torch.unsqueeze.html

```python
# torch.unsqueeze() - adds a single dimension
```

```python
print(f"Previous Tensor: {x_squeezed}")
print(f"Previous Shape: {x_squeezed.shape}")

# adds extra dimensions

x_unsqueezed = x_squeezed.unsqueeze(dim = 0)  # add the `1` at the 0th index dimension

print(f"\nNew tensor: {x_unsqueezed}")
print(f"New shape: {x_unsqueezed.shape}")
```

***torch.permute***: https://docs.pytorch.org/docs/2.11/generated/torch.permute.html

Rearrange the dimensions in a specific order!!

```python
x_original = torch.rand(size=(224, 224, 3)) # H W C channels

# permute the tensor to rearrage the axis

# convert from the HWC -> CWH
x_permuted = x_original.permute(2,0,1)
# shifts the 0->1, 1->2, 2->0
# x_original.shape, x_permuted.shape

print(f"previous shape: {x_original.shape}")
print(f"new shape: {x_permuted.shape}")
```
